```python
import pandas as pd

df = pd.read_csv("/content/GeoPulse_500_Synthetic_GPS_Pings.csv")


print(df.isnull().sum())

print(df.duplicated().sum())

invalid_lat = df[
    (df["latitude"] < -90) |
    (df["latitude"] > 90)
]

invalid_lon = df[
    (df["longitude"] < -180) |
    (df["longitude"] > 180)
]

df["timestamp"] = pd.to_datetime(
    df["timestamp"],
    errors="coerce"
)

print("Invalid latitude:", len(invalid_lat))
print("Invalid longitude:", len(invalid_lon))
print("Invalid timestamps:", df["timestamp"].isnull().sum())
```
